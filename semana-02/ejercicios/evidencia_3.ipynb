```python
import numpy
import matplotlib.pyplot

# --- 1. CARGA DE DATOS ---
data = numpy.loadtxt(fname='inflammation-01.csv', delimiter=',')
print("-> Datos cargados correctamente.")

# --- 2. EXPLORACIÓN ---
print("\n--- EXPLORACIÓN ---")
print("Tipo de estructura:", type(data))
print("Dimensiones de la matriz (Pacientes, Días):", data.shape)

# --- 3. ANÁLISIS ---
promedio_general = numpy.mean(data)
print("\n--- ANÁLISIS ---")
print("Inflamación promedio general:", promedio_general)

maximo_por_paciente = numpy.max(data, axis=1)
print("Inflamación máxima del primer paciente:", maximo_por_paciente[0])

# --- 4. VISUALIZACIÓN ---
print("\n--- VISUALIZACIÓN ---")
promedio_diario = numpy.mean(data, axis=0)

matplotlib.pyplot.plot(promedio_diario)
matplotlib.pyplot.title('Inflamación Promedio por Día')
matplotlib.pyplot.ylabel('Nivel de Inflamación')
matplotlib.pyplot.xlabel('Días del Estudio')
matplotlib.pyplot.show()
```

Al analizar el gráfico de la inflamación promedio diaria de los pacientes a lo largo de los 40 días del estudio, se observa lo siguiente:
Los niveles de inflamación comienzan bajos y aumentan de forma progresiva y prácticamente lineal hasta llegar al dia 20 donde se alcanza el pico máximo. Posterior a este día la inflamación promedio por día comienza a descender, por lo que se puede concluir que el tratamiento o medicación administrada tarda cerca de 20 días en hacer efecto.
